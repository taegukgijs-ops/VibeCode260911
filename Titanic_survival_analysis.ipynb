# 타이타닉 생존 데이터 분석

웹에서 타이타닉호 승객 데이터를 내려받아 기본 정보를 확인하고, 성별에 따른 생존율을 비교합니다.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_URL = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv'
titanic = pd.read_csv(DATA_URL)

print(f'데이터 크기: {titanic.shape[0]}행 x {titanic.shape[1]}열')
display(titanic.head())
```

## 데이터 확인

생존 여부(`survived`)와 성별(`sex`)을 중심으로 결측치와 주요 통계를 확인합니다.

```python
print('열별 결측치 수')
display(titanic.isna().sum().sort_values(ascending=False).to_frame('missing_count'))

print('생존 여부별 승객 수')
display(titanic['survived'].value_counts().rename(index={0: '사망', 1: '생존'}).to_frame('passenger_count'))
```

```python
survival_by_sex = (
    titanic.groupby('sex', as_index=False)['survived']
    .mean()
    .rename(columns={'sex': '성별', 'survived': '생존율'})
)
survival_by_sex['생존율(%)'] = survival_by_sex['생존율'] * 100
survival_by_sex['성별'] = survival_by_sex['성별'].map({'male': '남성', 'female': '여성'})

display(survival_by_sex[['성별', '생존율(%)']].round({'생존율(%)': 2}))
```

## 성별 생존율 바차트

성별별 생존율을 백분율로 표시합니다.

```python
plt.rcParams['font.family'] = ['Malgun Gothic', 'AppleGothic', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

fig, ax = plt.subplots(figsize=(7, 5))
bars = ax.bar(
    survival_by_sex['성별'],
    survival_by_sex['생존율(%)'],
    color=['#4C78A8', '#F58518'],
    width=0.55,
)

ax.set_title('성별에 따른 타이타닉 생존율', fontsize=15, pad=14)
ax.set_xlabel('성별')
ax.set_ylabel('생존율 (%)')
ax.set_ylim(0, 100)
ax.grid(axis='y', linestyle='--', alpha=0.35)
ax.set_axisbelow(True)

for bar in bars:
    rate = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        rate + 2,
        f'{rate:.1f}%',
        ha='center',
        va='bottom',
        fontsize=11,
    )

plt.tight_layout()
plt.show()
```
